````python

class Heap:

    def __init__(self):
        self.arreglo = [float('-inf')]

    def insert(self, valor):
        self.arreglo.append(valor)
        ih = len(self.arreglo) -1
        ip = ih //2
        while self.arreglo[ip] > self.arreglo[ih]:
            self.arreglo[ip], self.arreglo[ih] = self.arreglo[ih], self.arreglo[ip]
            ih = ip
            ip = ip // 2

    def remove_smallest(self):
        minimo = self.arreglo[1]
        valor = self.arreglo.pop()
        self.arreglo[1] = valor
        ip = 1
        while 2*ip < len(self.arreglo):
            im = 2*ip
            if 2*ip + 1 < len(self.arreglo) and self.arreglo[2*ip + 1] < self.arreglo[im]:
                im = 2*ip + 1
            if self.arreglo[ip] > self.arreglo[im]:
                self.arreglo[ip],self.arreglo[im] = self.arreglo[im],self.arreglo[ip]
                ip = im
            else:
                break

        return minimo
        
            

    def build_heap(self, lista):
        pass
````

## Para ver su calificación

````python
INF = float('-inf')

def nuevo_heap(Heap, valores):
    h = Heap()
    for v in valores:
        h.insert(v)
    return h

def test_1_insert_sin_subir(Heap):
    h = nuevo_heap(Heap, [10, 15, 20, 20, 50, 46, 30, 40])
    return h.arreglo == [INF, 10, 15, 20, 20, 50, 46, 30, 40]

def test_2_insert_sube_a_la_raiz(Heap):
    h = nuevo_heap(Heap, [10, 15, 20, 20, 50, 46, 30, 40, 5])
    return h.arreglo == [INF, 5, 10, 20, 15, 50, 46, 30, 40, 20]

def test_3_remove_retorna_minimo(Heap):
    h = nuevo_heap(Heap, [10, 15, 20, 20, 50, 46, 30, 40, 5])
    return [h.remove_smallest(), h.remove_smallest()] == [5, 10]

def test_4_remove_deja_heap(Heap):
    h = nuevo_heap(Heap, [10, 15, 20, 20, 50, 46, 30, 40, 5])
    for _ in range(3):
        h.remove_smallest()
    return h.arreglo == [INF, 20, 30, 20, 40, 50, 46]

def test_5_secuencia_completa(Heap):
    h = nuevo_heap(Heap, [10, 15, 20, 20, 50, 46, 30, 40, 5])
    for _ in range(3):
        h.remove_smallest()
    h.insert(3)
    h.insert(25)
    h.remove_smallest()
    h.remove_smallest()
    return h.arreglo == [INF, 20, 25, 40, 30, 50, 46]

TESTS = [test_1_insert_sin_subir, test_2_insert_sube_a_la_raiz, test_3_remove_retorna_minimo,
         test_4_remove_deja_heap, test_5_secuencia_completa]

sum([i(Heap) for i in TESTS])
````
