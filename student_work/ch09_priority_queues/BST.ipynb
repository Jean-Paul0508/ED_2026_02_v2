```python
def find_min(T: LinkedBinaryTree):
    '''retorna el contenedor donde se encuentra el valor mínimo'''
    current = T.root()
    while T.left(current) is not None:
        current = T.left(current)
    return current
```

```python
def es_ordenado(T: LinkedBinaryTree):
    '''retorna True si el árbol está ordenado, False en caso contrario'''
    it = T.inorder()

    try:
        anterior = next(it).element()
    except StopIteration:
        return True

    for pos in it:
        actual = pos.element()
        if actual < anterior:
            return False
        anterior = actual

    return True
```
